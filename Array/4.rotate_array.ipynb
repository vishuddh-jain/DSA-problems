### brute force method


```python
def rotate(nums, k) :
    if len(nums) == 1:
        return

    n = len(nums)

    if k > len(nums):
        k = k%n

    r = n-1
    l = n-k-1
        
    temp = nums[(n-k):]

    for i in range(n-k):
        nums[r] = nums[l]
        r-=1
        l-=1

    for i in range(k):
        nums[i] = temp[i]
    print(nums)

rotate(nums=[1,2,3,4,5,6,7],k=3)
```

### optimized version

```python
def rotate(nums, k) :
    n = len(nums)
    k = k % n   
    temp = nums[-k:] + nums[:-k]
    for i in range(len(temp)):
        nums[i] = temp[i]
    print(nums)

rotate(nums=[1,2,3,4,5,6,7],k=3)
```
